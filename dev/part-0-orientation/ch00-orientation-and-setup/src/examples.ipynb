## The first call pays for compilation

`solve_laplacian` is a jitted function. Its first call compiles the computation with XLA and runs it; the second
call only runs the compiled code. This is why the chapter says to time a GPU run twice.

```python
first, second = environment_check.time_twice(environment_check.solve_laplacian, 300)
print(f'first call  {first * 1e3:8.1f} ms  (compile + run)')
print(f'second call {second * 1e3:8.1f} ms  (run only)')
```

Try it yourself: call `time_twice` again with a different size (the function is recompiled for each new size,
because the size is a static argument) and with the same size (no recompilation: both calls are fast).

```python
print(environment_check.time_twice(environment_check.solve_laplacian, 300))
```
