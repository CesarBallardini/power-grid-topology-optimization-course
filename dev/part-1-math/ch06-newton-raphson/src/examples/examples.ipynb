## Newton on the 2 × 2 system

$F(x, y) = (x^2 + y^2 - 4,\ e^x + y - 1)$ from $(1, -1)$: each step solves $J\,\Delta = -F$. The exponent of
the residual roughly doubles each step, the signature of quadratic convergence.

```python
for k, (x, r) in enumerate(newton.newton(newton.residual, newton.jacobian, [1.0, -1.0])):
    print(k + 1, np.round(x, 6), f'{r:.1e}')
```

From $(-1, 1)$ Newton finds the other root; from $(0, 0)$ the Jacobian is singular. The lab's `newton_solve`
returns a success flag instead of raising.

```python
for x0 in ([-1.0, 1.0], [0.0, 0.0]):
    x, ok, hist = newton_lab.newton_solve(newton.residual, newton.jacobian, x0)
    print(f'from {x0}: success = {ok}, x = {np.round(x, 6)}, {len(hist) - 1} steps')
```

## Residual histories on a log scale

Quadratic convergence bends downwards on `semilogy`; a fixed damping of 0.5 gives a straight line (linear
convergence, the error halves each step).

```python
fig, ax = plt.subplots(figsize=(6, 4))
for label, kw in (('full steps', {}), ('damping 0.5', {'damping': 0.5}), ('backtracking', {'backtrack': True})):
    _, _, hist = newton_lab.newton_solve(newton.residual, newton.jacobian, [1.0, -1.0], **kw)
    ax.semilogy(hist, marker='o', label=label)
ax.set_xlabel('iteration')
ax.set_ylabel('‖F(x_k)‖')
ax.legend()
plt.show()
```

## Basins of attraction

Which root each starting point of $[-3, 3]^2$ reaches (dark: failure). The ragged boundary follows the curve
$\det J = 2x - 2y e^x = 0$, where the Newton step is unbounded.

```python
label = newton_lab.basins()
print(newton_lab.basin_counts(label))
s = np.linspace(-3, 3, 200)
fig, ax = plt.subplots(figsize=(5, 5))
ax.imshow(label, origin='lower', extent=(-3, 3, -3, 3), cmap='viridis')
ax.plot(s, s * np.exp(-s), 'r--', linewidth=1, label='det J = 0')
ax.set_ylim(-3, 3)
ax.legend()
plt.show()
```
