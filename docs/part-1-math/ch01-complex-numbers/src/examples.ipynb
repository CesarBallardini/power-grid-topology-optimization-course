## One load on a 230 V source

$Z = 40 + j30\ \Omega$, $V = 230\angle 0^\circ$ V. The current lags the voltage by the impedance angle, and
$S = V I^*$ gives $P$ and $Q$ at once.

```python
v, z = 230, 40 + 30j
i = phasor_power.load_current(v, z)
s = phasor_power.complex_power(v, i)
print('|Z|, angle Z :', abs(z), math.degrees(cmath.phase(z)))
print('I           :', i, '=', complex_plane.polar_deg(i))
print('S = V conj(I):', s, '  |S| =', abs(s), '  hypot(P, Q) =', math.hypot(s.real, s.imag))
```

## Multiplication is rotate and scale

Multiplying by $e^{j\theta}$ keeps the modulus and adds $\theta$ to the argument; multiplying two numbers
multiplies the moduli and adds the arguments.

```python
zz = 3 + 4j
for theta in (0, 30, 90, 180):
    w = complex_plane.rotate(zz, theta)
    print(f'theta = {theta:3d}  w = {w:.4f}  (abs, arg) = {complex_plane.polar_deg(w)}')
print(complex_plane.product_check(zz, 1 + 0.5j))
```

## Roots of unity

The three cube roots of 1 are $120^\circ$ apart and sum to zero: the balanced three-phase set of Chapter 10.

```python
roots = complex_plane.roots_of_unity(3)
print([f'{r:.4f}' for r in roots], ' sum =', f'{sum(roots):.1e}')
```

Try it yourself: change `3` to `5` or `12` and check that the sum is still zero, then rotate the whole set by a
common angle with `complex_plane.rotate` and check that the sum still vanishes.

```python
print(f'{sum(complex_plane.roots_of_unity(12)):.1e}')
```
